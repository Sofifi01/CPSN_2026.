```python
# Visualización para ciencia de datos / Recreando gráfico (Core)

import seaborn as sns
import matplotlib.pyplot as plt


# Aplicar el fondo y los colores de la imagen.
sns.set_theme(style="darkgrid")
sns.set_palette(["#E24A33", "#348ABD"])

# Cargar el dataset.
df = sns.load_dataset("tips")

# Crear la figura y el eje.
fig, ax = plt.subplots(figsize=(12, 6))

# El color representa el horario y la forma indica si es fumador.
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    hue_order=["Lunch", "Dinner"],
    style="smoker",
    style_order=["Yes", "No"],
    markers={"Yes": "o", "No": "X"},
    s=100,
    ax=ax
)

ax.set_title(
    "Relación entre Total Bill y Tip con Anotaciones y Flechas",
    fontsize=20
)
ax.set_xlabel("Total Bill ($)", fontsize=15)
ax.set_ylabel("Tip ($)", fontsize=15)

ax.legend(
    loc="upper right",
    fontsize="large",
    shadow=True,
    title="Detalles"
)

# Cuadrícula gris con líneas discontinuas.
ax.grid(
    True,
    which="major",
    linestyle="--",
    linewidth=0.5,
    color="gray"
)

# El texto queda fuera del eje, como en la imagen.
ax.annotate(
    "Mayor tip en la cena",
    xy=(50, 10),
    xytext=(40, 15),
    arrowprops=dict(
        facecolor="black",
        shrink=0.05
    )
)

# Flecha verde desde (30, 5) hacia (40, 8).
ax.arrow(
    30, 5,
    10, 3,
    head_width=0.8,
    head_length=0.5,
    fc="green",
    ec="green"
)

plt.show()
```
