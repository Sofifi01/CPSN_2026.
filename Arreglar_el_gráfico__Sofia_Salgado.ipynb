```python
# Visualización para ciencia de datos / Arreglar el gráfico (Core)

import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO


# 1. Datos de ejemplo.
# Cada registro debe estar en una línea distinta.
data = """fecha,ventas
2023-01-01,200
2023-01-02,220
2023-01-03,210
2023-01-04,205
2023-01-05,215
2023-01-06,225
"""


# 2. Cargar los datos.
# Usamos StringIO de la biblioteca io, en lugar de pd.compat.StringIO.
df = pd.read_csv(StringIO(data))


# 3. Convertir las fechas.
# El nombre correcto de la columna es 'fecha', no 'fechas'.
df["fecha"] = pd.to_datetime(df["fecha"], format="%Y-%m-%d")

# Ordenar las fechas para conectar los puntos en orden cronológico.
df = df.sort_values("fecha")

print(df)


# 4. Gráfico de líneas.
# Permite observar cómo cambian las ventas con el paso de los días.
plt.figure(figsize=(10, 6))
plt.plot(df["fecha"], df["ventas"], marker="o", color="steelblue")
plt.title("Ventas diarias")
plt.xlabel("Fecha")
plt.ylabel("Ventas")
plt.xticks(df["fecha"], rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# 5. Gráfico de dispersión.
# Creamos una figura independiente para este gráfico.
plt.figure(figsize=(10, 6))
plt.scatter(df["fecha"], df["ventas"], color="darkorange", s=70)
plt.title("Ventas diarias: gráfico de dispersión")
plt.xlabel("Fecha")
plt.ylabel("Ventas")
plt.xticks(df["fecha"], rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# 6. Histograma.
# Muestra cuántos días tienen ventas dentro de cada intervalo.
plt.figure(figsize=(10, 6))
plt.hist(df["ventas"], bins=5, color="skyblue", edgecolor="black")
plt.title("Distribución de las ventas diarias")
plt.xlabel("Ventas")
plt.ylabel("Frecuencia (cantidad de días)")
plt.yticks(range(0, len(df) + 1))
plt.tight_layout()
plt.show()
```
